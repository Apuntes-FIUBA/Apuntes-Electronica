```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import impulse, TransferFunction, step, bode


H_1 = TransferFunction([1, -1], [1, -1/2])

H_2 = TransferFunction([1, 0], [1, -3/2])

H = TransferFunction([4, -4, 0], [4, -8, 3])
```

```python
def rta_impulso(H):

    # Calcular la respuesta al impulso
    t, y = impulse(H)

    # Graficar la respuesta al impulso
    plt.figure(figsize=(8, 5))
    plt.plot(t, y, label='Respuesta al impulso')
    plt.title('Respuesta al Impulso de H(s)')
    plt.xlabel('Tiempo (s)')
    plt.ylabel('Amplitud')
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
def rta_escalon(H):
    # Calcular la respuesta al escalón
    t, y = step(H)

    # Graficar la respuesta al escalón
    plt.figure(figsize=(8, 5))
    plt.plot(t, y, label='Respuesta al escalón', color='blue')
    plt.title('Respuesta al Escalón de H(s)')
    plt.xlabel('Tiempo (s)')
    plt.ylabel('Amplitud')
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
def graf_bode(H):
    # Calcular el diagrama de Bode
    frequencies, magnitude, phase = bode(H)

    # Graficar la magnitud
    plt.figure(figsize=(10, 6))
    plt.subplot(2, 1, 1)
    plt.semilogx(frequencies, magnitude)  # Gráfica en escala logarítmica
    plt.title('Diagrama de Bode de H(s)')
    plt.ylabel('Magnitud (dB)')
    plt.grid(True, which='both', linestyle='--', linewidth=0.5)

    # Graficar la fase
    plt.subplot(2, 1, 2)
    plt.semilogx(frequencies, phase)  # Gráfica en escala logarítmica
    plt.xlabel('Frecuencia (rad/s)')
    plt.ylabel('Fase (grados)')
    plt.grid(True, which='both', linestyle='--', linewidth=0.5)

    plt.tight_layout()
    plt.show()
```

## $H_1(s)$

```python
rta_impulso(H_1)
rta_escalon(H_1)
graf_bode(H_1)
```

## $H_2(s)$

```python

rta_impulso(H_2)
rta_escalon(H_2)
graf_bode(H_2)
```

## $H(s)$

```python

rta_impulso(H)
rta_escalon(H)
graf_bode(H)
```
